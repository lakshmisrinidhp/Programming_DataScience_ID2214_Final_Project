```python
import pandas as pd 
import numpy as np

from rdkit import Chem #Chem can turn SMILES strings into molecule objects
from rdkit.Chem import rdMolDescriptors as rdDesc #it has a lot of functions that compute descriptors
from rdkit.Chem import Descriptors as Desc
from rdkit.Chem import Lipinski
```

```python
#columns with INDEX, SMILES, ACTIVE
train_df = pd.read_csv("training_smiles.csv")
#columns with INDEX and SMILES
test_df  = pd.read_csv("test_smiles.csv")

#show the first few rows of the table
train_df.head(), test_df.head()
```

```python
#convert SMILES to RDKit Mol objects

def smiles_to_mol(smiles: str): #take a string
    """Convert a SMILES string to an RDKIT Mol object, or None if it fails."""
    try:
        return Chem.MolFromSmiles(smiles) #parse the SMILES string and create a molecule object Mol
    except:
        return None #if fail
```

```python
train_df["mol"] = train_df["SMILES"].apply(smiles_to_mol)
test_df["mol"] = test_df["SMILES"].apply(smiles_to_mol)

print("Invalid mols in train:", train_df["mol"].isna().sum())
print("Invalid mols in test:", test_df["mol"].isna().sum())
```

```python
#Test cell
example_mol = train_df["mol"].iloc[0]
print("SMILES:", train_df["SMILES"].iloc[0])
print("Mol:", example_mol)
print("Type:", type(example_mol))
```

```python
descriptor_names = [
    "ExactMolWt",               #  1 exact molecular weight
    "NumRings",                 #  2 total number of rings
    "NumAromaticRings",         #  3 number of aromatic rings
    "NumAliphaticRings",        #  4 number of non-aromatic rings
    "NumHBD",                   #  5 H-bond donors
    "NumHBA",                   #  6 H-bond acceptors
    "TPSA",                     #  7 polar surface area
    "NumRotatableBonds",        #  8 number of rotatable bonds
    "HeavyAtomCount",           #  9 number of heavy atoms
    "NumHeteroatoms",           # 10 number of heteroatoms
    "FractionCSP3",             # 11 fraction of sp3 carbons
    "MolLogP",                  # 12 logP (Crippen)
    "MolMR",                    # 13 molar refractivity
    "NumValenceElectrons",      # 14 total valence electrons
    "NumSaturatedRings",        # 15 number of saturated rings
    "NumAromaticHeterocycles",  # 16 aromatic heterocycles
    "NumAromaticCarbocycles",   # 17 aromatic carbocycles
    "NumAliphaticHeterocycles", # 18 aliphatic heterocycles
    "NumAliphaticCarbocycles",  # 19 aliphatic carbocycles
    "LabuteASA",                # 20 Labute approximate surface area
   
]

len(descriptor_names)
```

```python
def compute_rdkit_descriptor_vector(mol):
    """
    Return a numpy array of 50 RDKit descriptors for one molecule.
    If mol is None (SMILES parsing failed), return a zero vector.
    """
    if mol is None:
        return np.zeros(len(descriptor_names), dtype=float)

    return np.array([
        rdDesc.CalcExactMolWt(mol),
        rdDesc.CalcNumRings(mol),
        rdDesc.CalcNumAromaticRings(mol),
        rdDesc.CalcNumAliphaticRings(mol),
        rdDesc.CalcNumHBD(mol),
        rdDesc.CalcNumHBA(mol),
        rdDesc.CalcTPSA(mol),
        rdDesc.CalcNumRotatableBonds(mol),
        Lipinski.HeavyAtomCount(mol),
        rdDesc.CalcNumHeteroatoms(mol),
        rdDesc.CalcFractionCSP3(mol),
        Desc.MolLogP(mol),
        Desc.MolMR(mol),
        Desc.NumValenceElectrons(mol),
        rdDesc.CalcNumSaturatedRings(mol),
        rdDesc.CalcNumAromaticHeterocycles(mol),
        rdDesc.CalcNumAromaticCarbocycles(mol),
        rdDesc.CalcNumAliphaticHeterocycles(mol),
        rdDesc.CalcNumAliphaticCarbocycles(mol),
        rdDesc.CalcLabuteASA(mol),
        Desc.NumRadicalElectrons(mol),
        Desc.NHOHCount(mol),
        Desc.NOCount(mol),
        Desc.BalabanJ(mol),
        Desc.BertzCT(mol),
        Desc.HallKierAlpha(mol),
        Desc.Kappa1(mol),
        Desc.Kappa2(mol),
        Desc.Kappa3(mol),
        Desc.Chi0(mol),
        Desc.Chi1(mol),
        Desc.Chi0v(mol),
        Desc.Chi1v(mol),
        Desc.Chi2v(mol),
        Desc.Chi3v(mol),
        Desc.Chi4v(mol),
        Desc.Chi0n(mol),
        Desc.Chi1n(mol),
        Desc.Chi2n(mol),
        Desc.Ipc(mol),
        Desc.MolWt(mol),
        Desc.HeavyAtomMolWt(mol),
        rdDesc.CalcNumSaturatedHeterocycles(mol),
        rdDesc.CalcNumSaturatedCarbocycles(mol),
        rdDesc.CalcNumAmideBonds(mol),
        rdDesc.CalcNumBridgeheadAtoms(mol),
        rdDesc.CalcNumSpiroAtoms(mol),
        Desc.PEOE_VSA1(mol),
        Desc.PEOE_VSA2(mol),
        Desc.PEOE_VSA3(mol),
    ], dtype=float)
```

```python
#test cell
compute_rdkit_descriptor_vector(train_df["mol"].iloc[0]), len(descriptor_names)
```

```python
#stack the small arrays into one big 2D array
X_train_desc = np.vstack(
    train_df["mol"].apply(compute_rdkit_descriptor_vector) #for each mol, call the descriptor function
)
y = train_df["ACTIVE"].values.astype(float)

X_train_desc.shape, y.shape
```

```python
X_test_desc = np.vstack(
    test_df["mol"].apply(compute_rdkit_descriptor_vector)
)

X_test_desc.shape
```

```python
np.save("X_train_rdkit_desc_50.npy", X_train_desc)
np.save("X_test_rdkit_desc_50.npy", X_test_desc)

pd.Series(descriptor_names).to_csv("rdkit_descriptor_names_20.txt", index=False)
```
